```python
import pandas as pd
Ames = pd.read_csv('Ames.csv')
# Tính toán tỷ lệ phần trăm giá trị bị thiếu cho mỗi cột
missing_data = Ames.isnull().sum()
missing_percentage = (missing_data / len(Ames)) * 100
data_type = Ames.dtypes
# Kết hợp số lượng và tỷ lệ phần trăm vào một DataFrame để dễ quan sát hơn
missing_info = pd.DataFrame({'Missing Values': missing_data,
'Percentage': missing_percentage,
'Data Type': data_type})

# Sắp xếp DataFrame theo tỷ lệ phần trăm giá trị bị thiếu theo thứ tự giảm dần
missing_info = missing_info.sort_values(by='Percentage', ascending=False)
# Hiển thị các cột có giá trị bị thiếu thuộc kiểu dữ liệu 'object'
print(missing_info[(missing_info['Missing Values'] > 0) &
(missing_info['Data Type'] == 'object')])
```

```python
import pandas as pd
import numpy as np
Ames = pd.read_csv('Ames.csv')
# Tính toán tỷ lệ phần trăm giá trị bị thiếu cho mỗi cột
missing_data = Ames.isnull().sum()
missing_percentage = (missing_data / len(Ames)) * 100
data_type = Ames.dtypes
# Kết hợp số lượng và tỷ lệ phần trăm vào một DataFrame để dễ quan sát hơn
missing_info = pd.DataFrame({'Missing Values': missing_data,
'Percentage': missing_percentage,
'Data Type': data_type})

# Sắp xếp DataFrame theo tỷ lệ phần trăm giá trị bị thiếu theo thứ tự giảm dần
missing_info = missing_info.sort_values(by='Percentage', ascending=False)
# Hiển thị các cột có giá trị bị thiếu thuộc kiểu dữ liệu số
print(missing_info[(missing_info['Missing Values'] > 0)
& (missing_info['Data Type'] == np.number)])
```

```python
# Khởi tạo một DataFrame để lưu trữ thông tin tóm tắt
concise_info = pd.DataFrame(columns=['Feature',

'Missing Values After Imputation',
'Mean Value Used to Impute'])

# Xác định và điền giá trị bị thiếu cho các biến số, đồng thời lưu trữ thông tin liên quan
missing_numeric_df = missing_info[(missing_info['Missing Values'] > 0) &
(missing_info['Data Type'] == np.number)]

for item in missing_numeric_df.index.tolist():
    mean_value = Ames[item].mean(skipna=True)
Ames[item].fillna(mean_value, inplace=True)
# Thêm thông tin vào DataFrame concise_info
concise_info.loc[len(concise_info)] = pd.Series({
'Feature': item,
'Missing Values After Imputation': Ames[item].isnull().sum(),
# Giá trị này phải là 0 vì tất cả các giá trị bị thiếu đã được điền
'Mean Value Used to Impute': mean_value
})
# Hiển thị concise_info DataFrame
print(concise_info)
```
