```python
import pandas as pd
import numpy as np
```

```python
# Tạo một DataFrame với các loại giá trị bị thiếu khác nhau
df = pd.DataFrame({
'A': [1, 2, np.nan, 4, 5],
'B': ['a', 'b', None, 'd', 'e'],
'C': [np.nan, np.nan, np.nan, np.nan, np.nan],
'D': [1, 2, 3, 4, 5]
})
# Sử dụng isnull() để xác định giá trị bị thiếu
missing_data = df.isnull().sum()
print(df)
print()
print(missing_data)
```
